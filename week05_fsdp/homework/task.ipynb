```python
import pickle
from torch.cuda._memory_viz import trace_plot
import os
```

```python
dirs = os.listdir("snapshots")

for d in dirs:
    with open(f"snapshots/{d}/rank0.pickle", "rb") as f:
        snapshot = pickle.load(f, encoding="utf-8")

    html = trace_plot(snapshot)

    with open(f"snapshots/{d}/rank0.html", "w") as f:
        f.write(html)
```
