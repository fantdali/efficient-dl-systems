```python
import torch
import run_epoch
import dataset
from run_epoch import DataMode
from transformers import AutoTokenizer
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
data_path = "wikitext-103-raw-v1/train-00000-of-00002.txt"
with open(data_path, "r") as f:
    train_data = f.read().split("\n = ")
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
import gc
gc.collect()
torch.cuda.empty_cache()
```

```python
torch.cuda.memory_allocated() / (1024**3)
```

```python
ds = dataset.BrainDataset(train_data, tokenizer)
```

```python
dataloader = torch.utils.data.DataLoader(ds, batch_size=8, shuffle=True, pin_memory=True) 
run_epoch.run_epoch(dataloader, tokenizer, device)
```
